```python
def  GetStockInfoByID(StocKID,strtype):
    try :
        result=data[data['公司代號']==StocKID]
        if strtype=='name':
            return result.iloc[0]['公司簡稱']
        if strtype=='category':
            if result.iloc[0]['Type']=="上櫃" :
                return result.iloc[0]['上櫃公司產業類別']
            else:
                return result.iloc[0]['上市公司產業類別']
        if strtype=='type':
            return result.iloc[0]['Type']
    except :
        print(StocKID)
```

```python
def stock_LevelArea(row):
    _level=float( row['level'])

    if row['interval_type']=='正區間': 
        _level=_level*1

    if row['interval_type']=='負區間':
        _level=_level*-1
    return _level

def stock_quote(row):
    try:
        open_price = float(row['開盤價'])
        if open_price == 0:
            return 0.0
        return round(float(row['漲跌價差']) / open_price * 100, 2)
    except (ValueError, TypeError):
        return 0.0
def  diff_quote(row):
    try :
        if float(row['now_price_y'])!=0:
            return format((  float(row['now_price_y']) - float(row['收盤價_x']) ) /float(row['now_price_y'])*100, '.2f')
        #if float(row['收盤價_y'])!=0:
            #return format((  float(row['收盤價_y']) - float(row['收盤價_x']) ) /float(row['收盤價_y'])*100, '.2f')
        # now_price_y 
    except FileNotFoundError:
        print(row['_quote'])

def trading_volume(row):
    data=row['成交股數']
    try :
         if row['Type2']=='上市': 
                data=data/1000
    except FileNotFoundError:
          print(data)
        
    return data
```

```python
def to_float(series):
    return pd.to_numeric(
        series.astype(str)
              .str.replace('%', '', regex=False)
              .str.replace(',', '', regex=False)
              .str.strip(),
        errors='coerce'
    )
```

```python
def calc_js_equivalent(df: pd.DataFrame, n: int = 5) -> pd.DataFrame:
    df = df.copy()
    df['日期_'] = pd.to_datetime(df['日期'])
    def generate_textdesc(row):
        price_desc = "漲" if row['漲跌BYN天Diff'] > 2 else "跌" if row['漲跌BYN天Diff'] < -3 else "-"
        volume_desc = "增" if row['增縮BYN天Diff'] > 2 else "縮" if row['增縮BYN天Diff'] < -10 else "-"
        return f"價{price_desc}量{volume_desc}"
    def _safe_int(x):
        try:
            return float(x)
        except (ValueError, TypeError):
            return np.nan
    def process_group(group):
        group = group.sort_values('日期_').reset_index(drop=True)
         # === 若沒有「量能均」，臨時用 n 天均量補上（不覆蓋原欄）===
        if '量能均' not in group.columns:
            group['_量能均_tmp'] = (
                pd.Series(group['量能']).apply(_safe_int).rolling(window=n, min_periods=1).mean()
            )
            ma_col = '_量能均_tmp'
        else:
            ma_col = '量能均'
    
        group['增縮BYN天'] = group['量能'].rolling(window=n).apply(lambda x: sum(x.dropna()), raw=False).fillna(0).astype(int) 
        group['增縮BYN天Diff'] = group['增縮BYN天'].rolling(window=2).apply(lambda x: sum(x.diff().dropna()), raw=False)
        
        group['漲跌BYN天'] = group['_quote'].rolling(window=n).apply(lambda x: sum(x.dropna()), raw=False).fillna(0).astype(int) 
        group['漲跌BYN天Diff'] = group['漲跌BYN天'].rolling(window=2).apply(lambda x: sum(x.diff().dropna()), raw=False)
            
        group['textdesc'] = group.apply(generate_textdesc, axis=1)
        
        #############################
          # === 對應你的 JS：量倍率（add - minus），以「日期由新到舊」挑最近符合條件的前 7 筆 ===
        g_desc = group.sort_values('日期_', ascending=False).copy()
      

        # 先計算 ratio = round(parseInt(量能)/parseInt(量能均), 2)
        vol_int = pd.to_numeric(g_desc['量能'].map(_safe_int), errors='coerce')
        #print(g_desc['量能'])
        vma_int = pd.to_numeric(g_desc[ma_col].map(_safe_int), errors='coerce')
        ratio = np.where(vma_int > 0, vol_int / vma_int, np.nan)
        # 模擬 toFixed(2) 後再 parseFloat
        ratio = np.round(ratio, 2)

        # add: _quote > 3 的前 7 筆 ratio 加總
        add = (
            pd.Series(ratio)[g_desc['_quote'] > 3].head(7).fillna(0).sum()
        )
        # minus: _quote < -2 的前 7 筆 ratio 加總
        minus = (
            pd.Series(ratio)[g_desc['_quote'] < -2].head(7).fillna(0).sum()
        )
        
        vol_multiplier = np.round(add - minus, 2)

        # 把這個分組層級的值寫回整組（和你的 JS 一樣是針對整個 datalist 求值）
        group['量倍率'] = float(vol_multiplier)

        # 清理臨時欄
        if '_量能均_tmp' in group.columns:
            group.drop(columns=['_量能均_tmp'], inplace=True)
        ##
        return group
    
    if 'stock_number' in df.columns:
        df = df.groupby('stock_number', group_keys=False).apply(process_group)
    else:
        df = process_group(df)

    return df
aa = calc_js_equivalent(aa, n=3)
```

```python
# ========== 產出 Result/summary.json ==========
colname = 'sectordetail'
analyze_ = aa.merge(detail_df, on='stock_number', how='left')
analyze_['sectordetail'] = analyze_['sector'] + '_' + analyze_['subsector']
analyze_['成交金額_補值'] = pd.to_numeric(analyze_['成交金額_補值'], errors='coerce')
analyze_['_quote_num'] = pd.to_numeric(analyze_['_quote'], errors='coerce')
analyze_['diff_quote_num'] = pd.to_numeric(analyze_['_quote'], errors='coerce')
analyze_['收盤價_num'] = pd.to_numeric(analyze_['收盤價'], errors='coerce')
analyze_['成交金額_num'] = pd.to_numeric(analyze_['成交金額'], errors='coerce')

summary = {}
all_dates = sorted(analyze_['日期'].unique().tolist())

for sector in analyze_[colname].dropna().unique():
    sec_df = analyze_[analyze_[colname] == sector]

    vol_s   = (sec_df.groupby('日期')['成交金額_補值'].sum() / 1_000_000).round(2)
    hot_s   = sec_df[sec_df['_quote_num'] > 6].groupby('日期')['stock_number'].count()
    total_s = sec_df.groupby('日期')['stock_number'].count()

    dates  = [d for d in all_dates if d in vol_s.index]
    vol    = [round(float(vol_s.get(d, 0)), 2) for d in dates]
    hot    = [int(hot_s.get(d, 0)) for d in dates]
    totals = [int(total_s.get(d, 0)) for d in dates]

    # rep_stocks：targetday 當日，成交金額最高前2檔
    today_sec = sec_df[sec_df['日期'] == targetday].copy()
    top2 = today_sec.nlargest(2, '成交金額_num')
    rep_stocks = []
    for _, row in top2.iterrows():
        rep_stocks.append({
            "id":    str(row['stock_number']),
            "name":  str(row['stock_name']),
            "price": float(row['收盤價_num']) if pd.notna(row['收盤價_num']) else 0.0,
            "quote": float(row['diff_quote_num']) if pd.notna(row['diff_quote_num']) else 0.0,
            "vol":   float(row['成交金額_num']) if pd.notna(row['成交金額_num']) else 0.0,
        })

    summary[str(sector)] = {
        "dates":       dates,
        "vol":         vol,
        "hot":         hot,
        "totals":      totals,
        "total":       len(sec_df['stock_number'].unique()),
        "stock_count": len(sec_df['stock_number'].unique()),
        "rep_stocks":  rep_stocks,
    }

with open('Result/summary.json', 'w', encoding='utf-8') as f:
    json.dump(summary, f, ensure_ascii=False, indent=2)

print(f"✅ summary.json 輸出完成，{len(summary)} 個產業，日期：{dates}")
```
