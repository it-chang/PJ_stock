```python
def  GetStockInfoByID(StocKID,strtype):
    try :
        result=data[data['公司代號']==StocKID]
        if strtype=='name':
            return result.iloc[0]['公司簡稱']
        if strtype=='category':
            if result.iloc[0]['Type']=="上櫃" :
                return result.iloc[0]['上櫃公司產業類別']
            else:
                return result.iloc[0]['上市公司產業類別']
        if strtype=='type':
            return result.iloc[0]['Type']
    except :
        print(StocKID)
```

```python
def calc_js_equivalent(df: pd.DataFrame, n: int = 5) -> pd.DataFrame:
    df = df.copy()
    df['日期_'] = pd.to_datetime(df['日期'])
    def generate_textdesc(row):
        price_desc = "漲" if row['漲跌BYN天Diff'] > 2 else "跌" if row['漲跌BYN天Diff'] < -3 else "-"
        volume_desc = "增" if row['增縮BYN天Diff'] > 2 else "縮" if row['增縮BYN天Diff'] < -10 else "-"
        return f"價{price_desc}量{volume_desc}"
    def _safe_int(x):
        try:
            return float(x)
        except (ValueError, TypeError):
            return np.nan
    def process_group(group):
        group = group.sort_values('日期_').reset_index(drop=True)
         # === 若沒有「量能均」，臨時用 n 天均量補上（不覆蓋原欄）===
        if '量能均' not in group.columns:
            group['_量能均_tmp'] = (
                pd.Series(group['量能']).apply(_safe_int).rolling(window=n, min_periods=1).mean()
            )
            ma_col = '_量能均_tmp'
        else:
            ma_col = '量能均'
    
        group['增縮BYN天'] = group['量能'].rolling(window=n).apply(lambda x: sum(x.dropna()), raw=False).fillna(0).astype(int) 
        group['增縮BYN天Diff'] = group['增縮BYN天'].rolling(window=2).apply(lambda x: sum(x.diff().dropna()), raw=False)
        
        group['漲跌BYN天'] = group['_quote'].rolling(window=n).apply(lambda x: sum(x.dropna()), raw=False).fillna(0).astype(int) 
        group['漲跌BYN天Diff'] = group['漲跌BYN天'].rolling(window=2).apply(lambda x: sum(x.diff().dropna()), raw=False)
            
        group['textdesc'] = group.apply(generate_textdesc, axis=1)
        
        #############################
          # === 對應你的 JS：量倍率（add - minus），以「日期由新到舊」挑最近符合條件的前 7 筆 ===
        g_desc = group.sort_values('日期_', ascending=False).copy()
      

        # 先計算 ratio = round(parseInt(量能)/parseInt(量能均), 2)
        vol_int = pd.to_numeric(g_desc['量能'].map(_safe_int), errors='coerce')
        #print(g_desc['量能'])
        vma_int = pd.to_numeric(g_desc[ma_col].map(_safe_int), errors='coerce')
        ratio = np.where(vma_int > 0, vol_int / vma_int, np.nan)
        # 模擬 toFixed(2) 後再 parseFloat
        ratio = np.round(ratio, 2)

        # add: _quote > 3 的前 7 筆 ratio 加總
        add = (
            pd.Series(ratio)[g_desc['_quote'] > 3].head(7).fillna(0).sum()
        )
        # minus: _quote < -2 的前 7 筆 ratio 加總
        minus = (
            pd.Series(ratio)[g_desc['_quote'] < -2].head(7).fillna(0).sum()
        )
        
        vol_multiplier = np.round(add - minus, 2)

        # 把這個分組層級的值寫回整組（和你的 JS 一樣是針對整個 datalist 求值）
        group['量倍率'] = float(vol_multiplier)

        # 清理臨時欄
        if '_量能均_tmp' in group.columns:
            group.drop(columns=['_量能均_tmp'], inplace=True)
        ##
        return group
    
    if 'stock_number' in df.columns:
        df = df.groupby('stock_number', group_keys=False).apply(process_group)
    else:
        df = process_group(df)

    return df
aa = calc_js_equivalent(aa, n=3)
```

```python
# ==================================================================
# checksun 分流與輸出（統一寫法 v2：makedirs + 輸出 + print 全部進 export_history）
# 依賴外部變數：aa, targetday, dateList, output_col, top5_index, to_float
# ==================================================================

# ---------- 設定 ----------
RESULT_DIR = 'Result'

EXCLUDED_TYPE1 = ['化學工業', '航運業', '生技醫療業', '塑膠工業', '金融保險', '鋼鐵工業', '居家生活']

HOLDINGS = ['1101', '1599', '1519', '2062', '2303', '2330', '2449', '6285', '6805', '8069']

# x 流：IG 推 https://www.instagram.com/moneyinallday/
WATCH_X = ["1519","2308","2342","2357","2379","2383","2401","2436","2454","2465","2498","2723","3006","3035","3138","3169","3217","3231","3374","3483","3526","3552","3661","3708","3715","4927","4968","4976","5243","5425","5457","5474","6223","6231","6271","6278","6438","6441","6515","6525","6533","6679","6689","6789","8069"]

# y 流：創高 http://stock.tendy.net/tdin_profit_rf.php?open=1
WATCH_Y = ["2408","6560","3030","2241","8086","6486","3530","5284","6841","3006","8046","4771","6274","1437","3380","6235","6584","6438","8432","3004","2455","1560","2419","5490","8109","1590","4161","3168","6690","2415","6112","5315","2637","3563","4128","1731","6491","4588","6158","4968","3564","4420","2402","6712","3141","6844","4153"]

# 要輸出成「分類資料夾」的標籤（其餘先停用）
EXPORT_LABELS = {
    '價_量': 'checksun_a',
    '【b1】': 'checksun_b1',
    '【b2】': 'checksun_b2',
    # '【c】':    'checksun_c',
    # '【d1】':   'checksun_d1',
    # '【d2】':   'checksun_d2',
    # '【e】':    'checksun_e',
    # '【x】看好': 'checksun_x',
      '【v】測試條件': 'checksun_v',
}


# ---------- 共用工具 ----------
def num(col):
    """所有數值比較統一入口：欄位 -> 數字 Series。"""
    return to_float(aa[col])


def stocks_where(cond):
    """符合條件的股票代號（去重）。"""
    return aa[cond]['stock_number'].drop_duplicates().tolist()


def export_history(stocks, folder, filename, df=None, verbose=True):
    """指定股票的完整歷史 -> Excel。
    一次處理：建資料夾 -> 去重/排序 -> 輸出 -> print(路徑, 股票數)。
    folder   : 資料夾（不存在會自動建立）
    filename : 檔名，例如 f'{targetday}.xlsx'
    df       : 預設 aa；RowData 版本傳 cc
    verbose  : 逐日迴圈想安靜時設 False
    """
    df = aa if df is None else df
    os.makedirs(folder, exist_ok=True)
    mask = df['stock_number'].isin(stocks)
    (df[output_col][mask]
        .drop_duplicates()
        .sort_values(by=['stock_number', '日期'], ascending=False)
        .to_excel(f'{folder}/{filename}', index=False))
    if verbose:
        print(f'{folder}/{filename}', len(set(stocks)))


def add_tag(condition, label):
    """僅在標籤尚未包含該類別時才加入。"""
    aa.loc[condition & (~aa['分類標籤'].str.contains(label, na=False, regex=False)), '分類標籤'] += f'{label};'


# ==================================================================
# Step 1：條件
# ==================================================================
is_today = (aa['日期'] == targetday)

avg_pct   = num('均價_%')
long_pct  = num('均價long_%')
macd_pct  = num('MACD_%')

# A：價量
cond_a = (
    (~aa['Type1'].isin(EXCLUDED_TYPE1))
    & aa['Full_Summary'].str.contains('多頭排列', na=False)
    & (num('成交股數') > 100)
    & (num('LevelArea') > 0)
    & ((num('Volume_MA_50') / num('成交金額平均').replace(0, np.nan)) > 0.5)
    & (num('營業利益率') > 0)
    & (num('累計營業收入-前期比較增減(%)') > 0)
    & (num('MACD_hist_diff') > 0)
    & (num('Volume_MA_5')  > num('Volume_MA_50') * 0.9)
    & (num('Volume_MA_10') > num('Volume_MA_50') * 0.9)
)

# B ~ E
cond_b1 = (avg_pct >= 0) & (long_pct > 0) & (macd_pct > 0)
cond_b2 = (avg_pct <  0) & (long_pct > 0) & (macd_pct > 0)
cond_c  =                  (long_pct > 0) & (macd_pct < 0)
cond_d1 = (avg_pct >= 0) & (long_pct < 0) & (macd_pct > 0)
cond_d2 = (avg_pct <  0) & (long_pct < 0) & (macd_pct > 0)
cond_e  =                  (long_pct < 0) & (macd_pct < 0)

# V
is_today = (aa['日期'] == targetday)

vol5  = num('Volume_MA_5')
vol10 = num('Volume_MA_10')
vol30 = num('Volume_MA_30')

# 現量5d比 = 成交金額 / Volume_MA_5（依你 JS 註解）
aa['現量5d比'] = (num('成交金額') / vol5.replace(0, np.nan)).round(2)
# 排序用欄位：now_price x 三大法人累計買賣超
aa['now_pricex三大法人_累計買賣超'] = num('now_price') * num('三大法人_累計買賣超')

cond_v = (
    is_today
    & (num('up_up_10d') > 0)
    & (vol5 > vol10)
    & (vol10 > vol30)
    & (aa['現量5d比'] > 1)
    & (vol5 > 10000)
)
# x / y
cond_x = is_today & aa['stock_number'].isin(WATCH_X)
cond_y = is_today & aa['stock_number'].isin(WATCH_Y)

# ==================================================================
# Step 2：標記分類標籤
# ==================================================================
aa['分類標籤'] = ''
for cond, label in [
    (cond_a,  '價_量'),
    (cond_b1, '【b1】'), (cond_b2, '【b2】'),
    (cond_c,  '【c】'),
    (cond_d1, '【d1】'), (cond_d2, '【d2】'),
    (cond_e,  '【e】'),
    (cond_x,  '【x】看好'),
    (cond_y,  '【y】創高'),
    (cond_v,  '【v】測試條件'),
]:
    add_tag(cond, label)
aa['分類標籤'] = aa['分類標籤'].str.strip(';')

# ==================================================================
# Step 3：依分類輸出（A ~ E / x / y）
# ==================================================================
for label, name in EXPORT_LABELS.items():
    stocks = stocks_where(aa['分類標籤'].str.contains(label, na=False, regex=False) & is_today)
    export_history(stocks, f'{RESULT_DIR}/{name}', f'{targetday}.xlsx')

# ==================================================================
# Step 4：F 分流（法人關注度 _quote > 4，逐日輸出）
# ==================================================================
'''
quote = num('_quote')
for _date in dateList:
    stocks = stocks_where((aa['日期'] == _date) & (quote > 4))
    export_history(stocks, f'{RESULT_DIR}/checksun_f', f'{_date}.xlsx')
'''
# ==================================================================
# S 分流：MA20*1.1 >= MA50（s 限強勢族群；sbk 為全部備份）
# ==================================================================
'''
aa['MA_20_int'] = num('MA_20').round().astype('Int64')
aa['MA_50_int'] = num('MA_50').round().astype('Int64')
cond_s = is_today & (aa['MA_20_int'] * 1.1 >= aa['MA_50_int'])

export_history(stocks_where(cond_s & aa['Typelevel'].isin(top5_index.tolist())),
               f'{RESULT_DIR}/checksun_s', f'{targetday}.xlsx')
export_history(stocks_where(cond_s),
               f'{RESULT_DIR}/checksun_sbk', f'{targetday}.xlsx')
'''

# ==================================================================
# SS 分流：近 6 日 MACD-SL 由負轉正，且當日 MACD_% >= 0
# ==================================================================
cond_ss_cross = ((num('MACD-SL') >= 0)
                 & (num('前日MACD-SL') <= 0)
                 & aa['日期'].isin(dateList[-6:]))
cond_ss_today = (macd_pct >= 0) & is_today

export_history(list(set(stocks_where(cond_ss_cross)) & set(stocks_where(cond_ss_today))),
               f'{RESULT_DIR}/checksun_ss', f'{targetday}.xlsx')

# ==================================================================
# 持有 / 分產業 / highlight
# ==================================================================
export_history(HOLDINGS, RESULT_DIR, 'checksunmy.xlsx')

for Type1 in sorted(aa['Type1'].astype(str).unique()):
    export_history(stocks_where(aa['Type1'] == Type1),
                   f'{RESULT_DIR}/checksun', f'{Type1}.xlsx', verbose=False)

#export_history(stocks_where(is_today & (aa['highlight'] == True)),
#               f'{RESULT_DIR}/checksun_h', f'{targetday}.xlsx')

# ==================================================================
# RowData 版本（空值補 0）：checkall / checkmy
# ==================================================================
cc = aa.copy()
cc[output_col] = cc[output_col].astype(str).replace(["NaN", "nan", "None", ""], "0")

for _date in dateList:
    day_df = cc[cc['日期'] == _date]
    export_history(day_df['stock_number'].unique().tolist(),
                   f'{RESULT_DIR}/checkall', f'{_date}.xlsx', df=day_df, verbose=False)

export_history(HOLDINGS, RESULT_DIR, 'checkmy.xlsx', df=cc[cc['日期'] == targetday])

# ==================================================================
# 量能遞增篩選(僅當日)
# ==================================================================
cols_v = output_col + ['現量5d比', 'now_pricex三大法人_累計買賣超']



stocks = stocks_where(cond_v)
export_history(stocks, f'{RESULT_DIR}/check_v', f'{targetday}.xlsx',
               df=aa[aa['日期'] == targetday])
```

```python
# ========== 產出 Result/summary.json ==========
colname = 'sectordetail'
analyze_ = aa.merge(detail_df, on='stock_number', how='left')
analyze_['sectordetail'] = analyze_['sector'] + '_' + analyze_['subsector']
analyze_['成交金額_補值'] = pd.to_numeric(analyze_['成交金額_補值'], errors='coerce')
analyze_['_quote_num'] = pd.to_numeric(analyze_['_quote'], errors='coerce')
analyze_['diff_quote_num'] = pd.to_numeric(analyze_['_quote'], errors='coerce')
analyze_['收盤價_num'] = pd.to_numeric(analyze_['收盤價'], errors='coerce')
analyze_['成交金額_num'] = pd.to_numeric(analyze_['成交金額'], errors='coerce')

summary = {}
all_dates = sorted(analyze_['日期'].unique().tolist())

for sector in analyze_[colname].dropna().unique():
    sec_df = analyze_[analyze_[colname] == sector]

    vol_s   = (sec_df.groupby('日期')['成交金額_補值'].sum() / 1_000_000).round(2)
    hot_s   = sec_df[sec_df['_quote_num'] > 6].groupby('日期')['stock_number'].count()
    total_s = sec_df.groupby('日期')['stock_number'].count()

    dates  = [d for d in all_dates if d in vol_s.index]
    vol    = [round(float(vol_s.get(d, 0)), 2) for d in dates]
    hot    = [int(hot_s.get(d, 0)) for d in dates]
    totals = [int(total_s.get(d, 0)) for d in dates]

    # rep_stocks：targetday 當日，成交金額最高前2檔
    today_sec = sec_df[sec_df['日期'] == targetday].copy()
    top2 = today_sec.nlargest(2, '成交金額_num')
    rep_stocks = []
    for _, row in top2.iterrows():
        rep_stocks.append({
            "id":    str(row['stock_number']),
            "name":  str(row['stock_name']),
            "price": float(row['收盤價_num']) if pd.notna(row['收盤價_num']) else 0.0,
            "quote": float(row['diff_quote_num']) if pd.notna(row['diff_quote_num']) else 0.0,
            "vol":   float(row['成交金額_num']) if pd.notna(row['成交金額_num']) else 0.0,
        })

    summary[str(sector)] = {
        "dates":       dates,
        "vol":         vol,
        "hot":         hot,
        "totals":      totals,
        "total":       len(sec_df['stock_number'].unique()),
        "stock_count": len(sec_df['stock_number'].unique()),
        "rep_stocks":  rep_stocks,
    }

with open('Result/summary.json', 'w', encoding='utf-8') as f:
    json.dump(summary, f, ensure_ascii=False, indent=2)

print(f"✅ summary.json 輸出完成，{len(summary)} 個產業，日期：{dates}")
```
