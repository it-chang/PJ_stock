# 上市公司基本資料

```python
# 【上市公司基本資料】
# https://data.gov.tw/dataset/18419

FALLBACK_PATHS = [
    r'BaseInfo/StockBaseData.csv',
    r'Result/StockBaseDataRead.csv',
]

def _load_fallback_listing():
    """從歷史備份檔讀取上市資料"""
    for path in FALLBACK_PATHS:
        try:
            for enc in ['utf-8', 'utf-8-sig', 'big5', 'cp950']:
                try:
                    df = pd.read_csv(path, encoding=enc, dtype=str)
                    df = df[df['Type'] == '上市'].copy()
                    print(f'[上市] 備援讀取成功：{path} ({enc})，共 {len(df)} 筆')
                    return df
                except (UnicodeDecodeError, LookupError):
                    continue
        except FileNotFoundError:
            continue
    print('[上市] 警告：備援檔案也找不到，df_listing2 為空 DataFrame')
    return pd.DataFrame(columns=['公司代號','公司名稱','公司簡稱','產業別','實收資本額','Type','上市公司產業類別','上櫃公司產業類別','新編碼'])

try:
    url = 'https://mopsfin.twse.com.tw/opendata/t187ap03_L.csv'
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.102 Safari/537.36"
    }
    req = requests.get(url, headers=headers, verify=False, timeout=15)
    req.raise_for_status()
    data = req.content.decode('utf8')
    if not data.strip() or '<html' in data.lower():
        raise ValueError('回傳內容非 CSV')
    df_listing = pd.DataFrame([x.replace('"', '').split(',') for x in data.split('\n')])
    df_listing = pd.DataFrame(df_listing.values[1:], columns=df_listing.iloc[0])
    df_listing['Type'] = '上市'
    df_listing1 = df_listing[['公司代號', '公司名稱', '公司簡稱', '產業別', '實收資本額', 'Type']]
    # 解碼產業 / Data 合併
    df_listing2 = pd.merge(df_listing1, mappingData, how='left', left_on='產業別', right_on='新編碼')
    print(f'[上市] 線上抓取成功，共 {len(df_listing2)} 筆')
except Exception as e:
    print(f'[上市] 線上抓取失敗（{e}），改讀備援檔...')
    df_listing2 = _load_fallback_listing()

df_listing2
```

# 上櫃公司基本資料

```python
# 【上櫃公司基本資料】
# UI: https://data.gov.tw/dataset/25036

def _load_fallback_oct():
    """從歷史備份檔讀取上櫃資料"""
    for path in FALLBACK_PATHS:
        try:
            for enc in ['utf-8', 'utf-8-sig', 'big5', 'cp950']:
                try:
                    df = pd.read_csv(path, encoding=enc, dtype=str)
                    df = df[df['Type'] == '上櫃'].copy()
                    print(f'[上櫃] 備援讀取成功：{path} ({enc})，共 {len(df)} 筆')
                    return df
                except (UnicodeDecodeError, LookupError):
                    continue
        except FileNotFoundError:
            continue
    print('[上櫃] 警告：備援檔案也找不到，dfOCT_ 為空 DataFrame')
    return pd.DataFrame(columns=['公司代號','公司名稱','公司簡稱','產業別','實收資本額','Type','上市公司產業類別','上櫃公司產業類別','新編碼'])

try:
    url = 'https://mopsfin.twse.com.tw/opendata/t187ap03_O.csv'
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.102 Safari/537.36"
    }
    req = requests.get(url, headers=headers, verify=False, timeout=15)
    req.raise_for_status()
    data = req.content.decode('utf8')
    if not data.strip() or '<html' in data.lower():
        raise ValueError('回傳內容非 CSV')
    dfOCT = pd.DataFrame([x.replace('"', '').split(',') for x in data.split('\n')])
    dfOCT = pd.DataFrame(dfOCT.values[1:], columns=dfOCT.iloc[0])
    dfOCT['Type'] = '上櫃'
    dfOCT1 = dfOCT[['公司代號', '公司名稱', '公司簡稱', '產業別', '實收資本額', 'Type']]
    # 解碼產業 / Data 合併
    dfOCT_ = pd.merge(dfOCT1, mappingData, how='left', left_on='產業別', right_on='新編碼')
    print(f'[上櫃] 線上抓取成功，共 {len(dfOCT_)} 筆')
except Exception as e:
    print(f'[上櫃] 線上抓取失敗（{e}），改讀備援檔...')
    dfOCT_ = _load_fallback_oct()

dfOCT_
```

# Output & 整合

```python
# output 上市上櫃股票資訊
try:
    _combined = pd.concat([df_listing2, dfOCT_])
    _combined.to_csv(r'BaseInfo/StockBaseData.csv', index=False, encoding='utf-8', errors='ignore')
    _combined.to_csv(r'Result/StockBaseDataRead.csv', index=False, encoding='big5', errors='ignore')
    print(f'[output] 儲存成功，共 {len(_combined)} 筆')
except Exception as e:
    print(f'[output] 儲存失敗：{e}')

_baseInfo = pd.concat([df_listing2, dfOCT_])
_baseInfo = _baseInfo[~_baseInfo['公司代號'].isna()]

df_info = _baseInfo.copy()
print(f'df_info 共 {len(df_info)} 筆，上市：{(df_info["Type"]=="上市").sum()}，上櫃：{(df_info["Type"]=="上櫃").sum()}')
df_info
```
