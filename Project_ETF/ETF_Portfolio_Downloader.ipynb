# 元大 ETF 投資組合下載器
自動抓取持股明細並輸出格式化 xlsx，目錄結構：
```
ETF_Investment_Portfolio/
├── 00981A/
│   ├── ETF_Investment_Portfolio_20260602.xlsx
│   └── ...
└── 00403A/
    └── ...
```
台灣時間 晚上6~7 點執行

````python
import os
# ========================================
# 設定：修改此處
# ========================================

ETF_LIST = [
    "00981A",
    "00403A",
    "0050",
    "0056",  # 如需其他 ETF 可在此加入
    "00878",
    "00713",
    "00929",
    "00919",
    "0052",
    
    "009819",
    "009805"
]

BASE_DIR = "ETF_Investment_Portfolio"  # 輸出根目錄
DAYS     = int(os.environ.get("ETF_DAYS", 10))  # 抓最近幾個交易日（沒開機的日子會補回來；最多約 120）
````

````python
import re
import requests
import pandas as pd
import time
from pathlib import Path
from datetime import datetime
from openpyxl import Workbook
from openpyxl.styles import Font, Alignment

today = datetime.today()
print(f"執行日期：{today.strftime('%Y-%m-%d')}（存檔用資料本身的日期）")
````

````python
# ========================================
# 抓取持股資料
# ========================================

URL = "https://www.pocket.tw/api/cm/MobileService/ashx/GetDtnoData.ashx"


def get_guest_token() -> str:
    """2026-10 起 API 要帶 token：從網頁內嵌的訪客 token 取得（約 1 天有效，每次執行重抓）。"""
    r = requests.get("https://www.pocket.tw/etf/tw/0050/fundholding",
                     headers={"User-Agent": "Mozilla/5.0"}, timeout=30)
    r.raise_for_status()
    m = re.search(r'tokens:\{at:"([^"]+)"', r.text)
    if not m:
        raise RuntimeError("網頁找不到訪客 token，pocket.tw 可能又改版了")
    return m.group(1)


TOKEN = get_guest_token()


def fetch_etf(etf: str) -> pd.DataFrame | None:
    """從 pocket.tw 抓取 ETF 最近 DAYS 個交易日的持股明細，回傳 DataFrame（含「日期」欄）。"""
    print(f"\n開始抓取 {etf}")
    params = {
        "action":    "getdtnodata",
        "DtNo":      "59449513",
        "ParamStr":  f"AssignID={etf};MTPeriod=0;DTMode=0;DTRange={DAYS};DTOrder=1;MajorTable=M722;",
        "FilterNo":  "0"
    }
    headers = {
        "User-Agent":    "Mozilla/5.0",
        "Referer":       f"https://www.pocket.tw/etf/tw/{etf}/fundholding",
        "Authorization": f"Bearer {TOKEN}",
    }
    try:
        r = requests.get(URL, params=params, headers=headers, timeout=30)
        r.raise_for_status()
        data = r.json()
        if "Title" not in data or "Data" not in data or len(data["Data"]) == 0:
            print(f"  {etf} 無資料 {data.get('Error', '')}")
            return None
        df = pd.DataFrame(data["Data"], columns=data["Title"])
        df["ETF"] = etf
        print(f"  完成 {etf} 共 {len(df)} 筆 | 日期 {df['日期'].min()} ~ {df['日期'].max()}")
        return df
    except Exception as e:
        print(f"  {etf} 抓取失敗: {e}")
        return None
````

````python
# ========================================
# 產生 xlsx（與範本格式完全相同）
# ========================================

def find_col(df: pd.DataFrame, candidates: list) -> str | None:
    """依候選清單找欄位名稱（容錯不同 ETF 欄位命名）。"""
    for c in candidates:
        if c in df.columns:
            return c
    return None


def write_xlsx(df: pd.DataFrame, path: Path, roc_date: str):
    """將持股 DataFrame 寫入格式化 xlsx（roc_date：民國資料日期）。"""
    wb = Workbook()
    ws = wb.active
    ws.title = "基金投資組合"

    # 欄寬
    ws.column_dimensions["A"].width = 24.75
    ws.column_dimensions["B"].width = 24.75
    ws.column_dimensions["C"].width = 12.9
    ws.column_dimensions["D"].width = 10.35

    bold12   = Font(name="Arial", bold=True,  size=12)
    normal12 = Font(name="Arial", bold=False, size=12)
    center   = Alignment(horizontal="center")
    left_a   = Alignment(horizontal="left")
    right_a  = Alignment(horizontal="right")

    def sc(coord, value, font=None, align=None):
        c = ws[coord]
        c.value = value
        if font:  c.font  = font
        if align: c.alignment = align

    # ── 資料日期
    sc("A1", f"資料日期：{roc_date}", font=bold12, align=left_a)

    # ── 基金資產（標題，合併列）
    sc("A3", "基金資產", font=bold12, align=center)
    ws.merge_cells("A3:C3")

    # ── 淨資產 / 流通單位 / 每單位淨值（空白佔位）
    for lc, lv, mr in [
        ("A4", "淨資產",       "B4:C4"),
        ("A5", "流通在外單位數", "B5:C5"),
        ("A6", "每單位淨值",    "B6:C6"),
    ]:
        sc(lc, lv, font=normal12, align=left_a)
        ws.merge_cells(mr)
        sc(mr.split(":")[0], "", font=normal12, align=right_a)

    # ── 基金資產表（期貨 / 股票）
    for coord, val in [("A8","項目"),("B8","金額"),("C8","權重")]:
        sc(coord, val, font=bold12, align=center)
    for coord, val, al in [
        ("A9",  "期貨(名目本金)", left_a),  ("B9",  "", right_a), ("C9",  "", right_a),
        ("A10", "股票",          left_a),  ("B10", "", right_a), ("C10", "", right_a),
    ]:
        sc(coord, val, font=normal12, align=al)

    # ── 現金部位表
    for coord, val in [("A12","項目"),("B12","金額"),("C12","權重")]:
        sc(coord, val, font=bold12, align=center)
    for i, label in enumerate(["現金","期貨保證金","申贖應付款","應收付證券款"]):
        r = 13 + i
        sc(f"A{r}", label, font=normal12, align=left_a)
        sc(f"B{r}", "",    font=normal12, align=right_a)
        sc(f"C{r}", "",    font=normal12, align=right_a)

    # ── 股票區（大標 + 表頭）
    sc("A19", "股票", font=bold12, align=center)
    ws.merge_cells("A19:D19")
    for coord, val in [("A20","股票代號"),("B20","股票名稱"),("C20","股數"),("D20","持股權重")]:
        sc(coord, val, font=bold12, align=center)

    # ── 持股明細（容錯欄位名）
    col_id   = find_col(df, ["標的代號","股票代號","代號"])
    col_name = find_col(df, ["標的名稱","股票名稱","名稱"])
    col_qty  = find_col(df, ["持有數","股數","數量"])
    col_wt   = find_col(df, ["權重(%)","持股權重","權重"])

    for excel_row, (_, row) in enumerate(df.iterrows(), start=21):
        val_id   = row[col_id]   if col_id   else ""
        val_name = row[col_name] if col_name else ""
        val_qty  = row[col_qty]  if col_qty  else ""
        val_wt   = row[col_wt]   if col_wt   else ""

        # 千分位格式
        try:
            val_qty = f"{int(str(val_qty).replace(',',''))  :,}"
        except: pass

        # 百分比格式（兩位小數）
        try:
            val_wt = f"{float(str(val_wt).replace('%','')):.2f}%"
        except: val_wt = str(val_wt)

        sc(f"A{excel_row}", str(val_id),   font=normal12, align=left_a)
        sc(f"B{excel_row}", str(val_name), font=normal12, align=left_a)
        sc(f"C{excel_row}", val_qty,        font=normal12, align=right_a)
        sc(f"D{excel_row}", val_wt,         font=normal12, align=right_a)

    path.parent.mkdir(parents=True, exist_ok=True)
    wb.save(str(path))
````

````python
# ========================================
# 主流程：抓取 → 依資料日期存 xlsx / 合併 CSV
# ========================================

def to_roc(yyyymmdd: str) -> str:
    return f"{int(yyyymmdd[:4]) - 1911}/{yyyymmdd[4:6]}/{yyyymmdd[6:]}"


base_dir = Path(BASE_DIR)
all_df   = []

for etf in ETF_LIST:
    df = fetch_etf(etf)
    if df is None:
        continue
    for d, g in df.groupby("日期"):
        write_xlsx(g, base_dir / etf / f"ETF_Investment_Portfolio_{d}.xlsx", to_roc(d))
    all_df.append(df)
    time.sleep(1)

if not all_df:
    raise RuntimeError("沒有成功取得任何資料")

# 合併 CSV（每個資料日期一檔）；這次沒抓到的 ETF 保留舊檔裡的資料
final_df = pd.concat(all_df, ignore_index=True)
done_etf = set(final_df["ETF"])
for d, g in final_df.groupby("日期"):
    final_csv = base_dir / f"ETF_HOLDING_ALL_{d}.csv"
    if final_csv.exists():
        old = pd.read_csv(final_csv, dtype=str)
        g = pd.concat([g, old[~old["ETF"].isin(done_etf)]], ignore_index=True)
    g.to_csv(final_csv, index=False, encoding="utf-8-sig")

dates = sorted(final_df["日期"].unique())
print(f"\n====================")
print(f"全部完成：{len(done_etf)}/{len(ETF_LIST)} 檔 ETF，資料日期 {dates[0]} ~ {dates[-1]}（{len(dates)} 天）")
missing = [e for e in ETF_LIST if e not in done_etf]
if missing:
    print("沒抓到：", missing)
````
