```python
import requests
import os

def download_csv_files(urls, folder="Result"):
    # 建立資料夾（若不存在）
    os.makedirs(folder, exist_ok=True)

    for url in urls:
        try:
            file_name = url.split("/")[-1]
            file_path = os.path.join(folder, file_name)

            res = requests.get(url)
            res.raise_for_status()

            with open(file_path, "wb") as f:
                f.write(res.content)

            print(f"✔ 下載完成: {file_name}")

        except Exception as e:
            print(f"✖ 下載失敗: {url} -> {e}")
```

```python
urls = [
    "https://mopsfin.twse.com.tw/opendata/t187ap05_L.csv",
    "https://mopsfin.twse.com.tw/opendata/t187ap05_O.csv",
    "https://mopsfin.twse.com.tw/opendata/t187ap05_R.csv"
]

download_csv_files(urls)
```
