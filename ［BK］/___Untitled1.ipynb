```python
today = datetime.today().date()
datalist = [
    (today - timedelta(days=i)).strftime("%Y-%m-%d")
    for i in range(15, -1, -1)
]

print(datalist)
```

```python
target_codes =_baseInfo.dropna()['公司代號'].iloc[::-1]
fund_result_list = []

for _date in datalist:
   
    try:
        fund_result = pd.read_excel(
             f'Result/fund/fund_result_{_date.replace("-", "")}.xlsx',
            engine='openpyxl'
        )
        # 只保留公司代號有在 target_codes 裡的資料
        fund_result['code'] = fund_result['code'].astype(str).str.strip()
        fund_result = fund_result[fund_result['code'].isin(target_codes)].copy()
        
        # 順便標記日期，方便之後區分是哪一天的資料
        fund_result['日期'] = _date
        fund_result_list.append(fund_result)
      
    except:
        print('error----->'+f'Result/fund/fund_result_{_date.replace("-", "")}.xlsx')

# 全部合併成一個 DataFrame
fund_result_all = pd.concat(fund_result_list, ignore_index=True)
```
